```python
features = [
    "SMOKING",
    "YELLOW_FINGERS",
    "ANXIETY",
    "PEER_PRESSURE",
    "CHRONIC DISEASE",
    "FATIGUE ",
    "ALLERGY ",
    "WHEEZING",
    "ALCOHOL CONSUMING",
    "COUGHING",
    "SHORTNESS OF BREATH",
    "SWALLOWING DIFFICULTY",
    "CHEST PAIN"
]

for feature in features:
    rate = pd.crosstab(
        df[feature],
        df["LUNG_CANCER"],
        normalize = "index"
    )*100 
print("\n",feature)
print(rate)
```

```python

results = []

for feature in features:
    rate = pd.crosstab(
        df[feature],
        df["LUNG_CANCER"],
        normalize="index"
    ) * 100
    
    results.append({
        "Feature": feature,
        "NO → Cancer YES %": rate.loc["NO", "YES"],
        "YES → Cancer YES %": rate.loc["YES", "YES"]
    })

feature_rates = pd.DataFrame(results)

feature_rates
```

```python
feature_rates["Difference"] = (
    feature_rates["YES → Cancer YES %"]
    - feature_rates["NO → Cancer YES %"]
)
```

```python
feature_rates.sort_values(
    "Difference",
    ascending=False
)
```

```python
plot_data = feature_rates.sort_values(
    "Difference",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    plot_data["Feature"],
    plot_data["Difference"]
)

plt.xlabel("Difference in Lung Cancer YES Rate (%)")
plt.ylabel("Feature")
plt.title("Feature Association with Lung Cancer")

plt.show()
```

```python
x = df.drop("LUNG_CANCER",axis=1)
y=df["LUNG_CANCER"]

y = y.map({
    "YES":1,
    "NO":0
})
print(x.shape)
print(y.shape)
print(y.value_counts())
```

```python
yes_no_columns = [
    "SMOKING",
    "YELLOW_FINGERS",
    "ANXIETY",
    "PEER_PRESSURE",
    "CHRONIC DISEASE",
    "FATIGUE ",
    "ALLERGY ",
    "WHEEZING",
    "ALCOHOL CONSUMING",
    "COUGHING",
    "SHORTNESS OF BREATH",
    "SWALLOWING DIFFICULTY",
    "CHEST PAIN"
]

x[yes_no_columns] = x[yes_no_columns].replace({
    "YES": 1,
    "NO": 0
})
```

```python
x["GENDER"] = x["GENDER"].map({
    "Male":1,
    "Female":0
})
```

```python
x.head()
```

```python
x.dtypes
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test, y_train , y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
stratify=y
)
```

```python
print("X_train:", x_train.shape)
print("X_test :", x_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training target:")
print(y_train.value_counts(normalize=True)*100)
print("\nTesting target:")
print(y_test.value_counts(normalize=True)*100)
```

```python
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(x_train_scaled, y_train)
```

```python
y_prob = model.predict_proba(x_test_scaled)
print(y_prob[:5])
```

```python
y_prob_yes = model.predict_proba(x_test_scaled)[:, 1]

print(y_prob_yes[:10])
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
comparison = pd.DataFrame({
    "Actual":y_test.values,
    "Predicted":y_pred,
    "Probability":y_prob_yes
})

comparison.head(10)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import confusion_matrix 
cm = confusion_matrix(y_test,y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_test, y_prob_yes)

print("ROC-AUC:", roc_auc)
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_prob_yes)

# Calculate AUC
roc_auc = roc_auc_score(y_test, y_prob_yes)

# Plot
plt.figure(figsize=(7, 5))

plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {roc_auc:.2f})")

# Random guessing line
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```
